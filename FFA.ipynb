```python
import json
from mealpy import FloatVar
from mealpy.swarm_based import *
import time
import numpy as np
```

```python
def Fun(solution):
  list1 = [0] * data['vmcount'][0]
  val = 0.0
  #print(solution)
  #print(int(round(solution[0])))
  for i in range(solution.size):
    solution[i]=round(solution[i])
    if(round(solution[i])!=0.0):
      list1[int(round(solution[i]))-1]=list1[int(round(solution[i]))-1] + data['task'][0]/data['VM'][int(round(solution[i]))-1]
    else:
      list1[int(round(solution[i]))]=list1[int(round(solution[i]))-1] + data['task'][0]/data['VM'][int(round(solution[i]))]
  #print(solution)
  return min(list1)
```

```python
f = open('/content/data.json')
data = json.load(f)
size = data['tcount'][0]
vmcount = data['vmcount'][0]

problem_dict = {
    "bounds": FloatVar(lb=[0 for i in range(size)], ub=[vmcount for i in range(size)], name="delta"),
    "obj_func": Fun,
    "minmax": "min",
    "verbose": True,
}

print("Task Count is set to", size)
print("VM Count is set to", vmcount)

start_time = time.time()
model =model = FFA.OriginalFFA(
    epoch=100,
    pop_size=100,
    gamma=0.001,
    beta_base=2,
    alpha=0.2,
    alpha_damp=0.99,
    delta=0.05,
    exponent=2
)

g_best = model.solve(problem_dict)

end_time = time.time()
print(f"Solution: {g_best.solution}, Fitness: {g_best.target.fitness}")
print(f"Solution: {model.g_best.solution}, Fitness: {model.g_best.target.fitness}")
```

```python
list=g_best.solution
s=np.round(g_best.solution, 0)
s = [str(int(i)) for i in s.tolist()]
exetime= end_time - start_time
print(exetime)
print(s)
data1 = {"name": s, "exetime": str(exetime)}

with open(f"FFA_sample_{size}_{vmcount}.json", "w") as outfile:
  json.dump(data1, outfile,ensure_ascii=False, indent=4)
```
